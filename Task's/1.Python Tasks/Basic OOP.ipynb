<h2> <b>Create a Student class with attributes for name, age,
and grades. Include methods to display details and
calculate the average grade.</b> <h2>

```python
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grades: {self.grades}")

    def calculate_average_grade(self):
        if not self.grades:
            return 0.0
        return sum(self.grades) / len(self.grades)


if __name__ == "__main__":

    student = Student(name="Ashish Mishra", age=20, grades=[85, 90, 78, 92])
    

    student.display_details()
    
    avg_grade = student.calculate_average_grade()
    print(f"Average Grade: {avg_grade:.2f}")
```
