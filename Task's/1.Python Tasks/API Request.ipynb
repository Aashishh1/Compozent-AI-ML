<h2><b>Use the requests library to fetch data from a public API
(e.g., JSONPlaceholder, Weather API, Quotes API). Parse
and display specific information.</b><h2>

```python
import requests

def fetch_and_display_quote(url):

    try:
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        if "quote" in data and "author" in data:
            quote = data['quote']
            author = data['author']
            print(f'"{quote}" - {author}')
        else:
            print(f"API response format not recognized. Expected 'quote' and 'author' fields.")

    except requests.exceptions.RequestException as e:
        print(f"An error occurred while fetching data: {e}")

user_provided_url = "https://quotes-api-self.vercel.app/quote"

fetch_and_display_quote(user_provided_url)
```
