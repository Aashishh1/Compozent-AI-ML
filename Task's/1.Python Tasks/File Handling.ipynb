<h2><b>Create a script to read a text file, count the number of
words, and write the results to a new file.</b><h2>

```python
def count_words(input_file, output_file):
    try:
        with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
            word_count = len(infile.read().split())
            outfile.write(f"Total number of words: {word_count}\n")
            print(f"Word count written to '{output_file}'.")
    except FileNotFoundError:
        print(f"Error: The file '{input_file}' was not found.")

input_file = "input.txt"
output_file = "output.txt"

with open(input_file, 'w') as file:
    file.write("Success usually comes to those who are too busy to be looking for it.")

count_words(input_file, output_file)

with open(output_file, 'r') as file:
    print(file.read())
```
